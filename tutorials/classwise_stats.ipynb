```python
BASELINE = "res_baseline_props"
RES = "res_wfgn_wfge_wgn_wprops_apool"
GATv2 = "gatv2_wfgn_wfge_wgn_wprops_apool"
GATv2_BASELINE_DIFF = "gatv2 - baseline "
RES_BASELINE_DIFF = "res - baseline"
```

### Summary statistics

Look at the overall performance:

```python
# mean improvement
mean_diff = df[[RES_BASELINE_DIFF, GATv2_BASELINE_DIFF]].mean()
print("Mean improvement over baseline:\n", mean_diff)
print()

# median improvement
median_diff = df[[RES_BASELINE_DIFF, GATv2_BASELINE_DIFF]].median()
print("Median improvement over baseline:\n", median_diff)
print()


# standard deviation of improvements
std_diff = df[[RES_BASELINE_DIFF, GATv2_BASELINE_DIFF]].std()
print("Variation in improvement:\n", std_diff)
```

### Count of labels improved / worsened

```python
res_improvement = (df[RES_BASELINE_DIFF] > 0).sum()
gat_v2_improvement = (df[GATv2_BASELINE_DIFF] > 0).sum()

print(f"Labels improved in ResGated Final model over baseline: {res_improvement}")
print(f"Labels improved in GATv2 Final model over baseline: {gat_v2_improvement}")
```

### Ranking per label

```python
df["best_model"] = df[[BASELINE, RES, GATv2]].idxmax(axis=1)
print(df["best_model"].value_counts())
```

```python
# Classes with highest difference
top_diff_res = df[RES_BASELINE_DIFF].sort_values(ascending=False).head(10)
print(
    f"\nTop 10 classes with highest improvement for {RES_BASELINE_DIFF}:\n",
    top_diff_res,
)

top_diff_gat = df[GATv2_BASELINE_DIFF].sort_values(ascending=False).head(10)
print(
    f"\nTop 10 classes with highest improvement for {GATv2_BASELINE_DIFF}:\n",
    top_diff_res,
)
```

```python
# Classes with minimal or no difference
minimal_diff_threshold = 0.01  # you can adjust
minimal_diff = df[df[RES_BASELINE_DIFF].abs() <= minimal_diff_threshold]
minimal_diff
```

```python
# Classes where baseline performs better
baseline_better = df[df[RES_BASELINE_DIFF] < 0]
print("\nClasses where baseline performs better than our model:\n", baseline_better)
```

```python
# Classes where both baseline and our model perform worse
# Assuming "worse" means below a threshold, e.g., f1 < 0.5
threshold = 0.5
both_worse = df[(df[BASELINE] < threshold) & (df[RES] < threshold)]
print(f"\nClasses where both baseline and {RES} have F1 < {threshold}:\n", both_worse)
```
