```python
import pandas as pd
```

```python
df = pd.read_csv("/home/aditya/python-chebai-graph/combined_f1_scores.csv")
df["aug-baseline"] = df["augmented"] - df["baseline"]
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
BASELINE = "baseline"
AUGMENTED = "augmented"
DIFF_COL = "aug-baseline"
```

### Summary statistics

Look at the overall performance:

```python
# mean improvement
mean_diff = df[[DIFF_COL]].mean()
print("Mean improvement over baseline:\n", mean_diff)
print()

# median improvement
median_diff = df[[DIFF_COL]].median()
print("Median improvement over baseline:\n", median_diff)
print()


# standard deviation of improvements
std_diff = df[[DIFF_COL]].std()
print("Variation in improvement:\n", std_diff)
```

### Count of labels improved / worsened

```python
gat_v2_improvement = (df[DIFF_COL] > 0).sum()

print(f"Labels improved in GATv2 Final model over baseline: {gat_v2_improvement}")
```

### Ranking per label

```python
# Classes with highest difference
top_diff_res = df[DIFF_COL].sort_values(ascending=False).head(10)
print(
    f"\nTop 10 classes with highest improvement for {DIFF_COL}:\n",
    top_diff_res,
)

```

```python
# Classes with minimal or no difference
minimal_diff_threshold = 0.01  # you can adjust
minimal_diff = df[df[DIFF_COL].abs() <= minimal_diff_threshold]
minimal_diff
```

```python
# Classes where baseline performs better
baseline_better = df[df[DIFF_COL] < 0]
print("\nClasses where baseline performs better than our model:\n", baseline_better)
```

```python
# Classes where both baseline and our model perform worse
# Assuming "worse" means below a threshold, e.g., f1 < 0.5
threshold = 0.5
both_worse = df[(df[BASELINE] < threshold) & (df[AUGMENTED] < threshold)]
print(f"\nClasses where both baseline and Augmented have F1 < {threshold}:\n", both_worse)
```

```python
# Classes where baseline performs better
baseline_better = df[df[DIFF_COL] < 0]
print("\nClasses where baseline performs better than our model:\n", baseline_better)
```
